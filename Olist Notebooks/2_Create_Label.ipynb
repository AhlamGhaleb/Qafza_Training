```python
import pandas as pd

ml_table = pd.read_parquet(
    "data/artifacts/01_ml_table.parquet"
)

print("Shape:", ml_table.shape)
```

```python
ml_table["order_delivered_customer_date"] = pd.to_datetime(
    ml_table["order_delivered_customer_date"]
)

ml_table["order_estimated_delivery_date"] = pd.to_datetime(
    ml_table["order_estimated_delivery_date"]
)
```

```python
labeled_orders = ml_table[
    ml_table["order_delivered_customer_date"].notna()
].copy()

print("عدد الطلبات ذات تاريخ التسليم:",
      len(labeled_orders))
```

```python
labeled_orders["is_delayed"] = (
    labeled_orders["order_delivered_customer_date"]
    > labeled_orders["order_estimated_delivery_date"]
).astype(int)
```

```python
check = labeled_orders[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "is_delayed"
    ]
].sample(10, random_state=42)

display(check)
```

```python
print("توزيع الفئات:")
print(labeled_orders["is_delayed"].value_counts())

print("\nالنسب المئوية:")
print(
    labeled_orders["is_delayed"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
labeled_orders.to_parquet(
    "data/artifacts/02_labeled_orders.parquet",
    index=False
)

print("تم حفظ 02_labeled_orders.parquet بنجاح.")
```
